```python
class Human:
    # コンストラクタを定義する
    def __init__(self, name, age):
        self.name = name
        self.age = age

    # 名前と年齢を表示するメソッドを定義する
    def printinfo(self):
        print("名前:", self.name)
        print("年齢:", self.age)

# インスタンス化する
attendee = Human("侍太郎", 35)

# メソッドを呼び出す
attendee.printinfo()
```
