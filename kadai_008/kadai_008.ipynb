```python
import random

# 変数varは正の整数（例として1～100）
var = random.randint(1, 100)

# 3と5の両方で割り切れる場合の処理を行う
if var % 3 == 0 and var % 5 == 0:
  print("FizzBuzz")
# 3で割り切れる場合の処理を行う
elif var % 3 == 0:
  print("Fizz")
# 5で割り切れる場合の処理を行う
elif var % 5 == 0:
  print("Buzz")
# どの条件にも該当しない場合
else:
  print(var)
```
