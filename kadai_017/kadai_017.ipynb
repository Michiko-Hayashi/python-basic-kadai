```python
class Human:
    # コンストラクタを定義する
    def __init__(self, name, age):
      self.name = name
      self.age = age

    # 成人判定メソッドを定義する。
    def check_adult(self):
      if self.age>=20:
          print(f"{self.name}は大人である")
      else:
          print(f"{self.name}は大人でない")

# インスタンスを作成する
human1 = Human("侍太郎", 30)
human2 = Human("侍二郎", 17)
human3 = Human("侍五郎", 8)
human4 = Human("侍一郎", 45)
human5 = Human("侍春", 14)

#リストに格納する
human_list = [human1, human2, human3, human4, human5]

# 繰り返し処理をする
for human in human_list:
  human.check_adult()
```
